```python
def solution(X, Y, D):
    distance = Y - X
    return (distance + D - 1) // D


print(solution(10, 85, 30))
```

```python
def solution(A):
    n = len(A)
    expected = (n + 1) * (n + 2) // 2
    return expected - sum(A)


print(solution([2, 3, 1, 5]))  # 4
print(solution([]))       
```

```python
def solution(A):
    total = sum(A)
    left = 0
    min_diff = float('inf')

    for value in A[:-1]:
        left += value
        right = total - left
        min_diff = min(min_diff, abs(left - right))

    return min_diff


print(solution([3, 1, 2, 4, 3]))  # 1
```
